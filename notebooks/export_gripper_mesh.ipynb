```python
meshes = []
fk = robot.visual_trimesh_fk(cfg=[0.04, 0.04])
for tm in fk:
    pose = fk[tm]
    mesh = tm.copy()
    mesh.apply_transform(pose)
    meshes.append(mesh)

```

```python
combined_mesh = trimesh.util.concatenate(meshes)
newpose = np.eye(4)
newpose[:3, 3] = np.array([0.1995, 0, 0]) # used by sai for viz
# newpose[:3, 3] = np.array([0.166, 0, 0]) # for palm behind aligned
combined_mesh.apply_transform(newpose)
```
